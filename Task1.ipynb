# Task 1: Iris EDA

**Goal**: Load, inspect, and visualize the Iris dataset to identify patterns and species separability using descriptive statistics and plotting.

# Key Insights
- **Species Separability**: Petal length and petal width show the highest separability between species.
- **Correlations**: Strong positive correlation exists between petal dimensions.
- **Data Integrity**: Clean dataset with no missing values.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# 1. Load Data
print("Loading Iris dataset...")
df = sns.load_dataset('iris')

# 2. Inspection (Required by prompt)
print("\n--- Dataset Info ---")
df.info()
print("\n--- Summary Statistics ---")
display(df.describe())

# 3. EDA/Visualization
sns.pairplot(df, hue='species', palette='husl')
plt.suptitle("Pairplot of Iris Features", y=1.02)
plt.show()

plt.figure(figsize=(12, 6))
plt.subplot(1, 2, 1)
sns.boxplot(x='species', y='sepal_length', data=df)
plt.title('Sepal Length by Species')

plt.subplot(1, 2, 2)
sns.boxplot(x='species', y='petal_length', data=df)
plt.title('Petal Length by Species')
plt.show()
```
