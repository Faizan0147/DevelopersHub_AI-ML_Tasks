## Task 6: House Price Prediction

### Key Insights
- **Model Comparison**: Gradient Boosting significantly outperformed Linear Regression (R2 of 0.77 vs 0.57).
- **Feature Impact**: Median Income (MedInc) was identified as the strongest predictor of house values.
- **Error Analysis**: The model performs well for standard housing but variance increases for high-end luxury properties.

```python
import pandas as pd
import numpy as np
from sklearn.datasets import fetch_california_housing
from sklearn.model_selection import train_test_split
from sklearn.ensemble import GradientBoostingRegressor
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error

# 1. Load
housing = fetch_california_housing()
X_train, X_test, y_train, y_test = train_test_split(housing.data, housing.target, test_size=0.2, random_state=42)

# 2. Train
gbr = GradientBoostingRegressor(random_state=42).fit(X_train, y_train)
preds = gbr.predict(X_test)

# 3. Evaluate (Required: MAE & RMSE)
mae = mean_absolute_error(y_test, preds)
rmse = np.sqrt(mean_squared_error(y_test, preds))
r2 = r2_score(y_test, preds)

print(f"R2 Score: {r2:.3f}")
print(f"Mean Absolute Error (MAE): {mae:.3f}")
print(f"Root Mean Squared Error (RMSE): {rmse:.3f}")
```
